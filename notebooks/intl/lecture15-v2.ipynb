### Problem 12

A student asked an AI: “Write a function that removes every word shorter than 3 letters from a list.”
The AI wrote the code below.
- (a) What does it actually print?
- (b) Explain the bug.
- (c) Fix it.

```python
def removeShort(words):
    for w in words:
        if len(w) < 3:
            words.remove(w)
    return words

print(removeShort(["a", "is", "cat", "to", "be", "dog"]))
```

### Problem 13

A student asked an AI: “Write a function that returns `True` if *every* row of a 2D list
contains the value `target`.” The AI wrote the code below.
- (a) What does it actually print?
- (b) Find a small input where the answer is wrong and explain why.
- (c) Fix it.

```python
def everyRowHas(g, target):
    for row in g:
        for x in row:
            if x == target:
                return True
    return False

print(everyRowHas([[1, 2], [2, 3]], 2))
print(everyRowHas([[1, 2], [3, 4]], 2))
```
